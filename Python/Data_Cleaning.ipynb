## Cell 1: Import libraries

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

## Cell 2: Load the dataset

```python
df = pd.read_csv("H:\Amazon Store Sales, Profitability & Returns Analytics\Dataset\Amazon Store Sales Data.xlsx - Sheet1 (1) (1).csv")
```

## Cell 4: Check column names

```python
df.columns
```

```python
df.rename(
    columns={"Row ID+O6G3A1:R6": "Row ID"},
    inplace=True
)
```

```python
df.columns
```

## Cell 5: Check data types and missing values

```python
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

## Cell 6: Check duplicates

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
```

## Cell 7: Convert dates

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce"
)
```

```python
df[["Order Date", "Ship Date"]].isnull().sum()
```

## Cell 8: Convert numerical columns


```python
numeric_columns = ["Sales", "Quantity", "Profit", "Returns"]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.info()
```

## Cell 9: Create Shipping Days

```python
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
df[["Order Date", "Ship Date", "Shipping Days"]].head()
```

## Cell 10: Create Return Status

```python
df["Return Status"] = np.where(
    df["Returns"] == 1,
    "Returned",
    "Not Returned"
)
```

## Cell 11: Create Profit Margin %

```python
df["Profit Margin %"] = np.where(
    df["Sales"] != 0,
    (df["Profit"] / df["Sales"]) * 100,
    0
)
```

## Cell 12: Create date features

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month_name()
df["Order Month No"] = df["Order Date"].dt.month
df["Order Quarter"] = "Q" + df["Order Date"].dt.quarter.astype(str)
```

## Cell 13: Check invalid values

```python
print("Negative Sales:", (df["Sales"] < 0).sum())
print("Negative Quantity:", (df["Quantity"] < 0).sum())
print("Negative Shipping Days:", (df["Shipping Days"] < 0).sum())
print("Missing Order Dates:", df["Order Date"].isnull().sum())
print("Missing Ship Dates:", df["Ship Date"].isnull().sum())
```

```python
df[df["Shipping Days"] < 0][
    ["Order ID", "Order Date", "Ship Date", "Shipping Days"]
]
```

## Cell 14: Final data-quality summary

```python
data_quality_summary = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Duplicate Rows",
        "Missing Values",
        "Negative Sales",
        "Negative Quantity",
        "Negative Shipping Days"
    ],
    "Value": [
        df.shape[0],
        df.shape[1],
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        (df["Sales"] < 0).sum(),
        (df["Quantity"] < 0).sum(),
        (df["Shipping Days"] < 0).sum()
    ]
})

data_quality_summary
```

## Cell 15: Save the clean dataset

```python
df.to_csv(
    "../Dataset/clean_amazon_store_sales.csv",
    index=False
)
```
